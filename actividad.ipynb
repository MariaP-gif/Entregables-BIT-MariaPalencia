```python
# 1. Registro de estudiantes
# ==========================================
estudiantes = [
    ("Ana", 20, 4.5),
    ("Carlos", 22, 3.8),
    ("Beatriz", 19, 4.2),
    ("David", 21, 2.5),
    ("Elena", 23, 4.8),
    ("Fernando", 20, 3.2),
    ("Gabriela", 18, 4.0),
    ("Hugo", 24, 2.8)
]

# ==========================================
# 2. Mostrar la información
# ==========================================
print("--- Información de estudiantes ---")
for nombre, edad, nota in estudiantes:
    print(f"{nombre} tiene {edad} años y obtuvo una nota de {nota}")

```

```python
# 3. Clasificación de estudiantes
# ==========================================
print("\n--- Clasificación por nota ---")
for nombre, edad, nota in estudiantes:
    if nota >= 4.5:
        clasificacion = "Excelente"
    elif nota >= 4.0:
        clasificacion = "Bueno"
    elif nota >= 3.0:
        clasificacion = "Aceptable"
    else:
        clasificacion = "Reprobó"

    print(f"{nombre}: {clasificacion}")
```

```python
# 4. Promedio general
# ==========================================
suma_notas = 0
for estudiante in estudiantes:
    suma_notas += estudiante[2]  # El índice 2 corresponde a la nota

promedio = suma_notas / len(estudiantes)
print(f"\nEl promedio general de las notas es: {promedio:.2f}")
```

```python
# 5. Búsqueda de estudiante
# ==========================================
print("\n--- Búsqueda de estudiante ---")
nombre_buscar = input("Ingresa el nombre del estudiante a buscar: ")

encontrado = False
for estudiante in estudiantes:
    # Se compara convirtiendo a minúsculas para evitar errores por mayúsculas
    if estudiante[0].lower() == nombre_buscar.lower():
        encontrado = True
        break

if encontrado:
    print(f"El estudiante '{nombre_buscar}' fue encontrado.")
else:
    print(f"No se encontró ningún estudiante con el nombre '{nombre_buscar}'.")
```

```python
# 6. Diccionario de ciudades
# ==========================================
print("\n--- Ciudades de origen ---")
ciudades_estudiantes = {
    "Ana": "Bogotá",
    "Carlos": "Medellín",
    "Beatriz": "Bogotá",
    "David": "Pereira",
    "Elena": "Bogotá",
    "Fernando": "Medellín",
    "Gabriela": "Cali",
    "Hugo": "Pereira"
}

for estudiante, ciudad in ciudades_estudiantes.items():
    print(f"{estudiante} vive en {ciudad}")
```

```python
# 7. Cantidad de estudiantes por ciudad
# ==========================================
conteo_ciudades = {}
for ciudad in ciudades_estudiantes.values():
    if ciudad in conteo_ciudades:
        conteo_ciudades[ciudad] += 1
    else:
        conteo_ciudades[ciudad] = 1

print("\nCantidad de estudiantes por ciudad:")
print(conteo_ciudades)
```

```python
# 8. Ciclo While
# ==========================================
print("\n--- Lectura de números (Ingrese 0 para terminar) ---")
cantidad = 0
suma_total = 0

while True:
    numero = float(input("Ingrese un número: "))
    if numero == 0:
        break
    cantidad += 1
    suma_total += numero

if cantidad > 0:
    promedio_numeros = suma_total / cantidad
    print(f"\nResultados del ciclo while:")
    print(f"• Cantidad de números ingresados: {cantidad}")
    print(f"• Suma total: {suma_total}")
    print(f"• Promedio: {promedio_numeros:.2f}")
else:
    print("\nNo se ingresaron números válidos antes del 0.")
```

```python
# 9. Uso de break y continue
# ==========================================
print("\n--- Recorrido del 1 al 30 con continue y break ---")
for num in range(1, 31):
    if num == 25:
        break
    if num % 3 == 0:
        continue
    print(num, end=" ")
print()  # Salto de línea final
```

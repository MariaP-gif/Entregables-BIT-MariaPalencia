```python
### Explicación de 3 problemas de calidad:

## 1. Redundancia y ruido por valores nulos masivos: La columna Animal code está totalmente vacía y Animal name presenta muchos valores nulos.

## 2. Duplicidad de registros: Existen 167 registros duplicados exactos generados durante la consolidación de fuentes de datos.

## 3. Falta de estandarización en variables categóricas: Las erratas tipográficas y el uso inconsistente de códigos/nombres completos impiden agrupar y analizar los datos de forma exacta.
```

```python
## LIMPIEZA DE DATOS ##
```

```python
# elimina duplicados
df = df.drop_duplicates()

df.shape
```

```python
## elimina columna sin informacion
df = df.drop(columns=['Animal code'])
```

```python
## Estandarización de variables categóricas
animal_mapping = {
    'European bison™': 'European bison', 'European bisson': 'European bison',
    'European buster': 'European bison', 'lynx?': 'lynx',
    'red squirel': 'red squirrel', 'red squirrell': 'red squirrel',
    'wedgehod': 'hedgehog', 'ledgehod': 'hedgehog'
}
country_mapping = {
    'PL': 'Poland', 'HU': 'Hungary', 'Hungry': 'Hungary',
    'CZ': 'Czech Republic', 'Czech': 'Czech Republic', 'DE': 'Germany'
}

df['Animal type'] = df['Animal type'].str.strip().replace(animal_mapping)
df['Country'] = df['Country'].str.strip().replace(country_mapping)
```

```python
df["Animal type"].value_counts()
```

```python
df["Country"].value_counts()
```

```python
# Filtrar países no pertenecientes al área de estudio de Europa Central/Oriental
df = df[~df['Country'].isin(['Australia', 'CC'])]
```

```python
# Conversión de tipos de datos
df['Observation date'] = pd.to_datetime(df['Observation date'], format='%d.%m.%Y', errors='coerce')
```

```python
# Imputación de nulos numéricos mediante la mediana por especie
df['Weight kg'] = df.groupby('Animal type')['Weight kg'].transform(lambda x: x.fillna(x.median()))
df['Body Length cm'] = df.groupby('Animal type')['Body Length cm'].transform(lambda x: x.fillna(x.median()))
```

```python
df.head()
```

```python
# valores faltantes existen por columna

df.isnull().sum()
```

```python
##1. Imputar Gender con la moda
df['Gender'] = df['Gender'].fillna(df['Gender'].mode()[0])
```

```python
## 2. Deducir Animal type por el más frecuente
df['Animal type'] = df['Animal type'].fillna(df['Animal type'].mode()[0])
```

```python
##3. Imputar Country con el país más frecuente
df['Country'] = df['Country'].fillna(df['Country'].mode()[0])
```

```python
# valores faltantes existen por columna

df.isnull().sum()
```

```python
## elimina columna por falta de mucha informacion
df = df.drop(columns=['Animal name'])
```

```python
## NUMPY

# Crear variable condicional de clasificación por tamaño
df['Es_Especie_Grande'] = np.where(df['Weight kg'] > 100, 'Grande', 'Pequeño/Mediano')

# Funciones estadísticas de NumPy sobre el peso
peso_promedio = np.nanmean(df['Weight kg'])
peso_mediana = np.nanmedian(df['Weight kg'])
peso_std = np.nanstd(df['Weight kg'])

print(f"\n[NumPy Stats] Promedio: {peso_promedio:.2f} kg | Mediana: {peso_mediana:.2f} kg | Desv. Estándar: {peso_std:.2f} kg")
```

```python
## ANÁLISIS CON GROUPBY()

# Análisis 1: Número de observaciones por tipo de animal
obs_por_animal = df.groupby('Animal type')['Country'].count()
print("\n--- 1. Observaciones por Tipo de Animal ---\n", obs_por_animal)

# Análisis 2: Promedio de peso y longitud por tipo de animal
promedios_animal = df.groupby('Animal type')[['Weight kg', 'Body Length cm']].mean()
print("\n--- 2. Promedio de Peso y Longitud por Animal ---\n", promedios_animal)

# Análisis 3: Promedio de peso por país
promedio_peso_pais = df.groupby('Country')['Weight kg'].mean()
print("\n--- 3. Promedio de Peso por País ---\n", promedio_peso_pais)

# Análisis 4: Conteo de observaciones por país y tipo de animal
obs_pais_animal = df.groupby(['Country', 'Animal type'])['Weight kg'].count()
print("\n--- 4. Conteo de Observaciones por País y Animal ---\n", obs_pais_animal)
```
